#SCHEMA ENFORCEMENT


```python
data = [(1,100,'parth',1),(2,200,'garvish',1),(3,300,'shivam',1),(4,400,'rudra',1)]
df = spark.createDataFrame(data,['id','salary','name','SUB_COL'])

display(df)
```

```python
df.write.format("delta")\
        .mode("append")\
        .save("/Volumes/deltalakeparth/default/deltavol/demosink/")
        
```
