#WORKING ON DELTA LOG

```python
data = [(1,100,'parth'),(2,200,'garvish'),(3,300,'shivam'),(4,400,'rudra')]
df = spark.createDataFrame(data,['id','salary','name'])

display(df)
```

WRITING IT IN VOLUME CREATED 

```python
df.write.format("delta")\
        .mode("append")\
        .save("/Volumes/deltalakeparth/default/deltavol/demosink/")
```

#READING DATA IN A PERTICULAR TRANSITION

```python
df = spark.read.format("delta").load(
    "/Volumes/deltalakeparth/default/deltavol/demosink/"
)

display(df)
```
