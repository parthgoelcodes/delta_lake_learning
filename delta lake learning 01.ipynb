##CREATING TABLE USING SQL 


```python
%sql
CREATE TABLE IF NOT EXISTS deltalakeparth.default.first_delta_table
(
    id INT,
    salary INT
)
USING DELTA
```

```python
DeltaTable.createIfNotExists(spark) \
  .tableName("deltalakeparth.default.first_spark_api") \
  .addColumn("id", "INT") \
  .addColumn("salary","INT")\
  .execute()
```
